```python
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

# Interactive 3D: 2 plots

```python
# Load the CSV file
file_path = 'data/fromMansur_fault_pressure_temp.csv'
data = pd.read_csv(file_path)

# Filter the data for fault_id=6
filtered_data = data[data['fault_id'] == 7]

# Extract necessary columns
xs_values = filtered_data['xs']
ys_values = filtered_data['ys']
zs_values = filtered_data['zs']
temps_values = filtered_data['temps']
pressures_values = filtered_data['pressures']

# Create subplots
fig = make_subplots(
    rows=1, cols=2,
    specs=[[{'type': 'scatter3d'}, {'type': 'scatter3d'}]],
    # subplot_titles=('Temperature Distribution', 'Pressure Distribution')
)

# Temperature plot
temp_scatter = go.Scatter3d(
    x=xs_values,
    y=ys_values,
    z=zs_values,
    mode='markers',
    marker=dict(
        size=2,
        color=temps_values,
        colorscale='plasma',
        colorbar=dict(
            title='Temperature (°C)',
            x=0.45,  # Positioning the color bar
            # xanchor='left'
        ),
        showscale=True
    ),
    name='Temperature'
)

# Pressure plot
pressure_scatter = go.Scatter3d(
    x=xs_values,
    y=ys_values,
    z=zs_values,
    mode='markers',
    marker=dict(
        size=2,
        color=pressures_values,
        colorscale='viridis',
        colorbar=dict(
            title='Pressure (Bar)',
            x=1.1,  # Positioning the color bar
            # xanchor='right'
        ),
        showscale=True
    ),
    name='Pressure'
)

# Add temperature scatter plot to subplot
fig.add_trace(temp_scatter, row=1, col=1)

# Add pressure scatter plot to subplot
fig.add_trace(pressure_scatter, row=1, col=2)

# Update layout
fig.update_layout(
    title={
        'text': 'Fault id = 6',
        'y':0.95,
        'x':0.5,
        'xanchor': 'center',
        'yanchor': 'top'
    },
    legend=dict(
        orientation='h',
        yanchor='bottom',
        y=1.1,
        xanchor='left',
        x=1.1
    ),
    scene=dict(
        xaxis_title='x (m)',
        yaxis_title='y (m)',
        zaxis=dict(
            title='z (m)',
            autorange='reversed' # reverse z-axis
        )
    ),
    scene2=dict(
        xaxis_title='x (m)',
        yaxis_title='y (m)',
        zaxis=dict(
            title='z (m)',
            autorange='reversed' # reverse z-axis
    )
    ),
    margin=dict(l=0, r=10, b=10, t=50)
)

# Show plot
fig.show()
```

# Interactive 3D: 4 plots

```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Load the CSV file
file_path = 'results_fault_slip_fault7.csv'
data = pd.read_csv(file_path)

# Filter the data for fault_id=6
filtered_data = data[data['faultName'] == 'fault7']

# Extract necessary columns
xs_values = filtered_data['x']
ys_values = filtered_data['y']
zs_values = filtered_data['z']
temps_values = filtered_data['temperature']
pressures_values = filtered_data['pressure']
sigmaThermal_values = filtered_data['sigmaThermal_MPa']
fault_slip_values = filtered_data['fault_slip']

# Create subplots
fig = make_subplots(
    rows=2, cols=2,
    specs=[[{'type': 'scatter3d'}, {'type': 'scatter3d'}],
           [{'type': 'scatter3d'}, {'type': 'scatter3d'}]]
    # subplot_titles=('Temperature', 'Pressure', 'Thermal stress', 'Fault Slip (0/1)')
)

# Temperature plot
temp_scatter = go.Scatter3d(
    x=xs_values,
    y=ys_values,
    z=zs_values,
    mode='markers',
    marker=dict(
        size=2,
        color=temps_values,
        colorscale='plasma',
        colorbar=dict(
            title='Temperature (°C)',
            x=0,  # Positioning the color bar
            y=0.75,
            len=0.5
        ),
        showscale=True
    ),
    name='Temperature'
)

# Pressure plot
pressure_scatter = go.Scatter3d(
    x=xs_values,
    y=ys_values,
    z=zs_values,
    mode='markers',
    marker=dict(
        size=2,
        color=pressures_values,
        colorscale='viridis',
        colorbar=dict(
            title='Pressure (Bar)',
            x=0.55,  # Positioning the color bar
            y=0.75,
            len=0.5
        ),
        showscale=True
    ),
    name='Pressure'
)

# Sigma Thermal plot
sigma_thermal_scatter = go.Scatter3d(
    x=xs_values,
    y=ys_values,
    z=zs_values,
    mode='markers',
    marker=dict(
        size=2,
        color=sigmaThermal_values,
        colorscale='cividis',
        colorbar=dict(
            title='Thermal stress (MPa)',
            x=0,  # Positioning the color bar
            y=0.15,
            len=0.5
        ),
        showscale=True
    ),
    name='Thermal stress'
)

# Fault Slip plot
fault_slip_scatter = go.Scatter3d(
    x=xs_values,
    y=ys_values,
    z=zs_values,
    mode='markers',
    marker=dict(
        size=2,
        color=fault_slip_values,
        colorscale='inferno',
        cmin=0,
        cmax=1,
        colorbar=dict(
            title='Fault Slip (yes/no)',
            x=0.55,  # Positioning the color bar
            y=0.15,
            len=0.5
        ),
        showscale=True
    ),
    name='Fault Slip'
)

# Add traces to subplots
fig.add_trace(temp_scatter, row=1, col=1)
fig.add_trace(pressure_scatter, row=1, col=2)
fig.add_trace(sigma_thermal_scatter, row=2, col=1)
fig.add_trace(fault_slip_scatter, row=2, col=2)

# Update layout to make the figure larger
fig.update_layout(
    title={
        'text': 'Fault id = 6',
        'y':0.98,
        'x':0.5,
        'xanchor': 'center',
        'yanchor': 'top'
    },
    legend=dict(
        orientation='h',
        yanchor='bottom',
        y=1,
        xanchor='right',
        x=1
    ),
    scene=dict(
        xaxis_title='x (m)',
        yaxis_title='y (m)',
        zaxis=dict(
            title='z (m)',
            autorange='reversed' # reverse z-axis
        )
    ),
    scene2=dict(
        xaxis_title='x (m)',
        yaxis_title='y (m)',
        zaxis=dict(
            title='z (m)',
            autorange='reversed' # reverse z-axis
    )),
    scene3=dict(
        xaxis_title='x (m)',
        yaxis_title='y (m)',
        zaxis=dict(
            title='z (m)',
            autorange='reversed' # reverse z-axis
    )),
    scene4=dict(
        xaxis_title='x (m)',
        yaxis_title='y (m)',
        zaxis=dict(
            title='z (m)',
            autorange='reversed' # reverse z-axis
    )),
    margin=dict(l=0, r=0, b=0, t=50),
    height=650,  # Set the height of the figure
    width=1000   # Set the width of the figure
)

# Show plot
fig.show()
```

# Static 3D plot

```python
# Load the CSV file
file_path = 'fromMansur_fault_pressure_temp.csv'
data = pd.read_csv(file_path)

# Filter the data for fault_id=10
filtered_data = data[data['fault_id'] == 1]

# Extract necessary columns
is_values = filtered_data['is']
js_values = filtered_data['js']
ks_values = filtered_data['ks']
temps_values = filtered_data['temps']

# Create a 3D plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot data
sc = ax.scatter(is_values, js_values, ks_values, c=temps_values, cmap='viridis')
plt.colorbar(sc, label='Temperature')

# Set labels
ax.set_xlabel('IS')
ax.set_ylabel('JS')
ax.set_zlabel('KS')
ax.set_title('3D Plot of Temperatures vs (IS, JS, KS) for fault_id=10')

# Show plot
plt.show()
```
